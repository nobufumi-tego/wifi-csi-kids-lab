```python
import numpy as np
import matplotlib.pyplot as plt
```

## 1. Wavelength = speed ÷ frequency (1-2)

---

## 1. 波長 ＝ 速さ ÷ 周波数（1-2）

```python
c = 299_792_458                      # speed of light, m/s / 光の速さ
for f_hz in [2.4e9, 2.462e9, 5.0e9]:  # try other frequencies! / ほかの周波数もためそう
    wavelength_m = c / f_hz
    print(f"{f_hz / 1e9} GHz -> {wavelength_m * 100:.2f} cm")
```

## 2. Draw a sine wave (1-2)

A wave goes up and down again and again. `np.sin` draws that shape.

---

## 2. サイン波をかく（1-2）

波は上がったり下がったりをくり返します。`np.sin` がその形をかいてくれます。

```python
x = np.linspace(0, 3, 600)            # position, in wavelengths / 場所（波長いくつ分）
plt.plot(x, np.sin(2 * np.pi * x))
plt.xlabel("position (wavelengths)")
plt.ylabel("height")
plt.show()
```

## 3. Path length S → person → R (1-3)

S = (0, 0), R = (3, 0), person at (1.5, y). `np.hypot(a, b)` = $\sqrt{a^2 + b^2}$ (Pythagoras).

---

## 3. S → 人 → R の道の長さ（1-3）

S = (0, 0)、R = (3, 0)、人は (1.5, y)。`np.hypot(a, b)` は $\sqrt{a^2 + b^2}$（ピタゴラスの定理）。

```python
y_m = np.linspace(0, 2, 201)          # distance from the straight line / 線からのきょり
path_m = 2 * np.hypot(1.5, y_m)
plt.plot(y_m, path_m)
plt.xlabel("person's distance from the line y [m]")
plt.ylabel("path S -> P -> R [m]")
plt.show()
print("y = 1.0 m ->", round(2 * np.hypot(1.5, 1.0), 4), "m")
```

## 4. Add two waves (1-4)

Change `shift` to 0, 0.25, 0.5, 1.0. When is A + B biggest? When does it almost vanish?

---

## 4. 2 つの波を足す（1-4）

`shift` を 0、0.25、0.5、1.0 に変えてみましょう。A + B がいちばん大きいのは？ ほとんど消えるのは？

```python
shift = 0.5                                  # extra path, in wavelengths / 長い分（波長いくつ分）
wave_a = np.sin(2 * np.pi * x)
wave_b = np.sin(2 * np.pi * (x - shift))
plt.plot(x, wave_a, label="path A")
plt.plot(x, wave_b, label="path B")
plt.plot(x, wave_a + wave_b, label="A + B", linewidth=3)
plt.legend()
plt.xlabel("position (wavelengths)")
plt.show()
```

## 5. Move the person and watch the received signal (1-4)

The receiver gets the straight wave (size 1) plus the wave bounced off the person
(size 0.5). As the person moves, the size of the sum goes up and down.

---

## 5. 人を動かして、受け取る電波を見る（1-4）

受信機は、まっすぐな波（大きさ 1）と、人ではね返った波（大きさ 0.5）の合計を受け取ります。
人が動くと、合計の大きさが上がったり下がったりします。

```python
wavelength_m = c / 2.462e9                   # channel 11 / チャンネル 11
extra_m = 2 * np.hypot(1.5, y_m) - 3.0       # bounced path is longer by this much / 長い分
turns = extra_m / wavelength_m               # in wavelengths / 波長いくつ分
received = np.abs(1 + 0.5 * np.exp(-2j * np.pi * turns))  # size of the sum / 合計の大きさ
plt.plot(y_m, received)
plt.xlabel("person's distance from the line y [m]")
plt.ylabel("size of the received wave")
plt.show()
```
