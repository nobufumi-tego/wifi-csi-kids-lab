Make a pretend recording with the simulator and look at it in three ways:

Run each cell with **Shift + Enter**.

# 01 · CSI を見てみよう
シミュレーターで記録を作り、3 つの見方で見てみます。表、いくつかのサブキャリアの時間変化、ヒートマップです。
ハードウェアは要りません。**Shift + Enter** でセルを 1 つずつ実行しましょう。

```python
%matplotlib inline
from csi_lab import SCENARIOS, simulate
from csi_lab.plot import plot_overview, plot_subcarriers

for name, meaning in SCENARIOS.items():
    print(f"{name:8s} {meaning}")
```

## 1. A recording is a table
One row = one Wi-Fi packet (about 100 per second). Columns: time, signal strength (RSSI),
label, and the amplitude on 56 subcarriers (`sc-28` … `sc28`).

---

## 1. 記録は表になっている
1 行 ＝ Wi-Fi のパケット 1 つ（1 秒に約 100 個）。列は、時刻、信号の強さ（RSSI）、ラベル、
そして 56 本のサブキャリアの振幅（`sc-28` 〜 `sc28`）です。

```python
walk = simulate("walk", duration_s=30, seed=1)
df = walk.to_dataframe()
print(walk.n_packets, "packets,", round(walk.duration_s, 1), "seconds,", round(walk.rate_hz, 1), "per second")
df.head()
```

## 2. A few subcarriers over time
Compare someone walking with someone sitting still. Which one wobbles more?

---

## 2. いくつかのサブキャリアの時間変化
歩いている人と、座って静かにしている人をくらべます。どちらがよくゆれていますか？

```python
still = simulate("still", duration_s=30, seed=1)
plot_subcarriers(walk).suptitle("walk")
plot_subcarriers(still).suptitle("still");
```

## 3. The heat map: all subcarriers at once
Top: time → right, subcarrier ↑, color = relative amplitude. Stripes mean something moved.
Middle: one motion number per second. Bottom: RSSI.

---

## 3. ヒートマップ: 全部のサブキャリアをいっぺんに
上: 右へ時間、上へサブキャリア、色 ＝ 相対的な振幅。しま模様は何かが動いたしるしです。
まん中: 1 秒ごとの動きの数。下: RSSI。

```python
plot_overview(walk, title="walk")
plot_overview(still, title="still");
```
